```python
import numpy as np
import matplotlib.pyplot as plt

# Load your .npy file
data = np.load("mario_sequences_augmented.npy")

# Select one sample (e.g., the first one)
sequence_sample = data[0]  # shape: (50, 7)

# Create the table image
fig, ax = plt.subplots(figsize=(60, 50))
ax.axis('off')

# Set column headers
column_labels = [f"Feature {i}" for i in range(sequence_sample.shape[1])]

# Create the table
tbl = ax.table(cellText=sequence_sample, colLabels=column_labels, loc='center')

# Make it readable
tbl.auto_set_font_size(False)
tbl.set_fontsize(10)
tbl.scale(2.0, 2.0)

# Save as image
plt.tight_layout()
plt.savefig("npy_sample_sequence_0.png", dpi=300)
plt.close()
```
