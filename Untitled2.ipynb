```python
import numpy as np
import matplotlib.pyplot as plt

# Example: load your .npy file
data = np.load("mario_sequences_augmented.npy")  # Replace with your actual file

# Take first 10 rows (or any slice you want)
sample = data[:10]

# Create a larger figure for more space
fig, ax = plt.subplots(figsize=(100, 100))  # Wider and taller canvas

# Hide axes
ax.axis('off')

# Create the table
tbl = ax.table(cellText=sample, loc='center')

# Optional: improve readability
tbl.auto_set_font_size(False)  # So you can control font size manually
tbl.set_fontsize(10)           # Increase font size
tbl.scale(2, 2)                # Scale the entire table (width, height)

plt.tight_layout()
plt.savefig("npy_sample_table.png", dpi=300)
plt.close()
```
