```python
import pandas as pd
import matplotlib.pyplot as plt

# Load your CSV
df = pd.read_csv("combined_trajectories.csv")

# Sample rows 489 to 499 (inclusive)
sample = df.iloc[489:500]

# Plot and save as image
fig, ax = plt.subplots(figsize=(20, 4))  # Optional: Adjust size for readability
ax.axis('off')  # Hide axes
tbl = ax.table(cellText=sample.values, colLabels=sample.columns, loc='center')

# Optional: Style the table
tbl.auto_set_font_size(False)
tbl.set_fontsize(8)
tbl.scale(1.2, 1.2)

plt.tight_layout()
plt.savefig("csv_sample_489_499.png", dpi=300)
plt.close()
```
